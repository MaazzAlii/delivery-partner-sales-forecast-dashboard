## Notebook Objectives:
1. Load forecasts from Moving Average (`moving_average_forecast.csv`) and Linear Regression (`regression_forecast.csv`)
2. Compare test set performance metrics (MAE, MAPE) side-by-side
3. Plot comparative actual vs predicted trajectories
4. Formulate executive verdict for non-technical restaurant owners
5. Export canonical champion forecast feed `model/outputs/final_forecast.csv`

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_ma = pd.read_csv('../model/outputs/moving_average_forecast.csv')
df_reg = pd.read_csv('../model/outputs/regression_forecast.csv')

# Evaluate test set (months 18-23)
test_actuals = df_ma.iloc[18:24]['actual_orders'].values
ma_preds = df_ma.iloc[18:24]['predicted_orders'].values
reg_preds = df_reg.iloc[18:24]['predicted_orders'].values

ma_mape = np.mean(np.abs((test_actuals - ma_preds) / test_actuals)) * 100
reg_mape = np.mean(np.abs((test_actuals - reg_preds) / test_actuals)) * 100

print(f"SMA-6 Baseline MAPE: {ma_mape:.2f}%")
print(f"Linear Regression MAPE: {reg_mape:.2f}%")
```

## Executive Verdict
**Winning Model**: Multiple Linear Regression

**Business Rationale**: Linear Regression achieves a 2.75% MAPE (~51 orders/month average error) compared to 7.96% MAPE (~148 orders/month error) for Moving Average. Linear Regression captures demand signals from Ramadan, rainy weather, and marketing promotions, enabling accurate inventory and staffing planning.
