## Notebook Objectives:
1. Load cleaned dataset (`data/careem_partner_monthly_clean.csv`)
2. Implement 3-month (SMA-3) and 6-month (SMA-6) simple moving average models
3. Backtest models on the final 6 historical months
4. Compute MAE & MAPE error metrics and select the optimal window
5. Generate out-of-sample 3-month forward forecasts and save to `model/outputs/moving_average_forecast.csv`

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

clean_data_path = "../data/careem_partner_monthly_clean.csv"
df = pd.read_csv(clean_data_path)
df.head()
```

```python
# Backtesting function
def evaluate_sma(window):
    backtest_indices = list(range(len(df) - 6, len(df)))
    actuals = df.loc[backtest_indices, 'orders'].values
    preds = [df.loc[idx-window:idx-1, 'orders'].mean() for idx in backtest_indices]
    mae = np.mean(np.abs(actuals - preds))
    mape = np.mean(np.abs((actuals - preds) / actuals)) * 100
    return mae, mape, preds

mae_3, mape_3, preds_3 = evaluate_sma(3)
mae_6, mape_6, preds_6 = evaluate_sma(6)

print(f"SMA-3 -> MAE: {mae_3:.2f}, MAPE: {mape_3:.2f}%")
print(f"SMA-6 -> MAE: {mae_6:.2f}, MAPE: {mape_6:.2f}%")
```

```python
# Generate 3-month forward forecast
best_window = 3 if mape_3 <= mape_6 else 6
last_orders = df['orders'].tolist()
fut_preds = []
for _ in range(3):
    pred = np.mean(last_orders[-best_window:])
    fut_preds.append(pred)
    last_orders.append(pred)

print(f"3-Month Forward Forecast (SMA-{best_window}): {fut_preds}")
```
